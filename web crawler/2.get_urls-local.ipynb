```python
from web_clawer_local import crawler
```

```python
htmlpath="./nkust.html"
url="https://ws1.nkust.edu.tw/Activity/"
skipfile=r"./files/skip/nkust-skip.txt"
out="./files/urls/nkust.txt"
```

```python
a=crawler(htmlpath,url)
a.add_skip_title(skipfile)

a.start()
    
hrefs=a.get_hrefs()
f=open(out,mode='w')
print(len(hrefs))
for i in hrefs:
    f.write(i+"\n")
    #webbrowser.open(i, new=2)
    print(i)
f.close()
```
