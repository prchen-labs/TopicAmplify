```python
from selenium import webdriver #enable javascript on the website
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from pyvirtualdisplay import Display
from bs4 import BeautifulSoup

max_href=1000
hrefs=[]
url=r"https://ws1.nkust.edu.tw/Activity/"
skipfile=open("./files/skip/nkust-skip.txt", mode="w", encoding="utf-8")

display = Display(visible=0, size=(1920, 1080))  
display.start()
chrome_options = Options()
#chrome_options.add_argument('--headless')
chrome_options.add_argument('--incognito')
ser = Service(ChromeDriverManager().install())
browser = webdriver.Chrome(service=ser,options=chrome_options)# or add to your PATH
browser.get(url)

soup = BeautifulSoup(browser.page_source, "html.parser")
result = soup.find_all("a",limit=max_href)
titles=[]
for i in result:
    title = i.get_text()
    if title is not None:
        if title not in titles :
            titles.append(title)
            skipfile.write(title)
            skipfile.write("\n")
        
skipfile.close()
```
