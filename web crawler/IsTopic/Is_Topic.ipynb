```python
import os
from ckipnlp.pipeline import CkipPipeline, CkipDocument
pipeline = CkipPipeline(opts={'con_parser': {'username': os.environ["CKIP_USERNAME"], 'password': os.environ["CKIP_PASSWORD"]}})
from numpy import zeros
```

```python
def flatten(l):
    return [item for sublist in l for item in sublist]
```

```python
import json
dictionary = []
with open("files/topics_dictionary2.txt", mode="r") as f: 
    jsonn = json.loads(f.read())
for i in range(len(jsonn)):
    dictionary.append((jsonn[i][0],i+1))
#print(dictionary)
```

```python
words=[]
for info in content:
    doc = CkipDocument(raw = info)
    pipeline.get_ws(doc)
    words += flatten(doc.ws.to_list())
words_index=[]
for word in words:
    for d in dictionary:
        if word == d[0]:
            words_index.append(d[1])
            break
    else:
        words_index.append(0)
```

```python
print(words_index)
for i in words_index:
    if i == 0:
        print("?",end="")
    else:
        print(dictionary[i-1][0],end="")
print()
```

```python
import numpy as np
results = np.zeros((1, len(dictionary)+1))
for i in words_index:
    results[0][i] = 1
```

```python
import tensorflow as tf
model = tf.keras.models.load_model('files/model')
print(model.predict(results))
```
