```python
import tensorflow_decision_forests as tfdf
import pandas as pd

# Load a dataset in a Pandas dataframe.
train_df = pd.read_csv("files/is-topic/trainning_set/train.csv")
test_df = pd.read_csv("files/is-topic/trainning_set/test.csv")
predict_df = pd.read_csv("files/is-topic/trainning_set/predict.csv")

# Convert the dataset into a TensorFlow dataset.
train_ds = tfdf.keras.pd_dataframe_to_tf_dataset(train_df, label="class")
test_ds = tfdf.keras.pd_dataframe_to_tf_dataset(test_df, label="class")
predict_ds = tfdf.keras.pd_dataframe_to_tf_dataset(predict_df)

# Train a Random Forest model.
model = tfdf.keras.RandomForestModel()
model.fit(train_ds)
```

```python
predict_ds
```

```python
# Summary of the model structure.
model.summary()
```

```python
# Evaluate the model.
model.evaluate(test_ds)
```

```python
# Export the model to a SavedModel.
model.save("files/is-topic/model")
```

```python
from keras.saving.saving_api import save_model
save_model(model,"files/is-topic/model")
```

```python
# Predict the model.
r=model.predict(predict_ds)
print(r)
```
