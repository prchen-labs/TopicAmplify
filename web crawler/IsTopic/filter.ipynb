```python
import json

path = "../../LDA Topic Model/files/dictionary.txt"
newpath = "../../LDA Topic Model/files/dictionary2.txt"

with open(path, mode="r") as f: 
    jsonn = json.loads(f.read())
    
data = []
for i in jsonn:
    if i[1] < 10:
        continue
    data.append(i)
    
with open(newpath, mode="w") as f: 
    f.write(json.dumps(data, indent=4, ensure_ascii=False))
```
