```python
from web_clawer import crawler
import webbrowser
```

```python
url=r"https://www.yuntech.edu.tw/index.php/2019-07-25-11-32-12/2019-04-10-08-06-44/itemlist?start={}"
start=0
page=57
add=16
skipfile=r"./files/skip/nyust-skip.txt"
out="./files/urls/nyust.txt"
```

```python
a=crawler(url,page = 0)
a.add_skip_title(skipfile)

for i in range(start,page+1):
    print(i)
    a.start()
    u=a.next_page(add)
    print(u)
    
hrefs=a.get_hrefs()
f=open(out,mode='w')
print(len(hrefs))
for i in hrefs:
    f.write(i+"\n")
    #webbrowser.open(i, new=2)
    print(i)
f.close()
```
