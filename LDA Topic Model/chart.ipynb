```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np
```

# Define the input array

```python
array = [[0, 3, 3, 8, 2], [1, 3, 1, 1, 7], [1, 6, 5, 3, 4], [2, 0, 0, 0, 4], [2, 0, 0, 0, 7], [2, 4, 0, 3, 5], [2, 4, 1, 3, 4], [2, 4, 5, 2, 4], [2, 6, 5, 3, 4], [2, 7, 5, 3, 4], [2, 8, 5, 2, 4], [4, 1, 8, 0, 2], [4, 3, 1, 0, 5], [4, 4, 1, 3, 7], [4, 7, 1, 1, 6], [5, 1, 1, 3, 7], [5, 3, 0, 5, 8], [5, 6, 0, 4, 5], [5, 7, 5, 5, 3], [6, 3, 4, 6, 4], [6, 5, 1, 6, 5], [7, 0, 5, 8, 5], [7, 3, 8, 1, 7], [7, 4, 1, 4, 3], [7, 8, 1, 4, 3], [8, 0, 1, 4, 5], [8, 2, 7, 2, 7], [8, 3, 1, 0, 5], [8, 4, 8, 5, 5], [8, 5, 4, 6, 4]]
```

```python
# define the array
arr = np.array(array)

# create the figure and the axes
fig, ax = plt.subplots()

# plot the array as a heatmap
im = ax.imshow(arr, cmap='viridis')

# set the ticks and tick labels for the axes
ax.set_xticks(np.arange(arr.shape[1]))
ax.set_yticks(np.arange(arr.shape[0]))
ax.set_xticklabels(np.arange(1, arr.shape[1]+1))
ax.set_yticklabels(np.arange(1, arr.shape[0]+1))

# rotate the x-axis labels and show them in the middle of the cell
plt.setp(ax.get_xticklabels(), rotation=0, ha="center", rotation_mode="anchor")

# loop over the data and annotate each cell with its value
for i in range(arr.shape[0]):
    for j in range(arr.shape[1]):
        text = ax.text(j, i, arr[i, j], ha="center", va="center", color="w")

# add a colorbar
cbar = ax.figure.colorbar(im, ax=ax)

# set the title of the chart
ax.set_title("2D Array Chart")

# show the chart
plt.show()
```

```python
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

x_data, y_data, z_data = [], [], []
for row in array:
    for point in row:
        if len(point) == 1:
            x_data.append(point[0])
            y_data.append(row.index(point))
            z_data.append(array.index(row))
        else:
            for sub_point in point:
                x_data.append(sub_point)
                y_data.append(row.index(point))
                z_data.append(array.index(row))

ax.scatter(x_data, y_data, z_data)

ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.set_zlabel('Z Axis')

plt.show()
```
