```python
import os

def AddPath(env,path):
    envstr = os.getenv(env)
    if envstr != None :
        if path not in envstr:
            os.environ[env] = envstr + ":" + path
    else:
         os.environ[env] =  path
            
AddPath("PATH","/usr/local/cuda/bin")
AddPath("CPATH","/usr/local/cuda/include")
AddPath("LD_LIBRARY_PATH","/usr/local/cuda/lib64")
AddPath("LD_LIBRARY_PATH","/usr/lib/x86_64-linux-gnu")
AddPath("LD_LIBRARY_PATH","/home/jupyter-selab/.local/lib/python3.9/site-packages/tensorrt/")

print(os.getenv("PATH"))
print(os.getenv("CPATH"))
print(os.getenv("LD_LIBRARY_PATH"))
```

```python
from ckipnlp.pipeline import CkipPipeline, CkipDocument
```

```python
pipeline = CkipPipeline(opts={'con_parser': {'username': os.environ["CKIP_USERNAME"], 'password': os.environ["CKIP_PASSWORD"]}})
```

```python
info="""為提升本校專利質量與加速業務流程，說明本校專利補助業務流程與注意事項，由成功大學企業關係與技轉中心林意茵經理主講優化機制及表單說明。"""
```

```python
doc = CkipDocument(raw=info)#info is 文本

# Word Segmentation
pipeline.get_ws(doc)
print(doc.ws)
for line in doc.ws:
    print(line.to_text())
```

```python
# Part-of-Speech Tagging
pipeline.get_pos(doc)
print(doc.pos)
for line in doc.pos:
    print(line.to_text())
```

```python
# Named-Entity Recognition
pipeline.get_ner(doc)
print(doc.ner)
```

```python
# Constituency Parsing
pipeline.get_conparse(doc)
print(doc.conparse)
```

```python
from ckipnlp.container.util.wspos import WsPosParagraph

# Word Segmentation & Part-of-Speech Tagging
for line in WsPosParagraph.to_text(doc.ws, doc.pos):
    print(line)
```
