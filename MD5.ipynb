```python
aaa=[]
for i in range(10):
    aaa.append(chr(48+i))
for i in range(26):
    aaa.append(chr(65+i))
for i in range(26):
    aaa.append(chr(97+i))
print(aaa)
```

```python
import hashlib
import random
```

```python
def ran(n):
    #val="0183c@God_To_A540_50@02_07_10@A545_ToA445_10@"
    #val="002f9@God_To_A843_50@02_07_11@A445_To_A999_3@"
    val="002f9@God_To_A265_50@02_07_11@A445_To_A999_3@"
    ans="00"
    print(" Process %d started.\n" % n,end='')
    while 1+1==2:
        b=""
        for i in range(5):
            b+=aaa[random.randint(0,len(aaa)-1)]
        val+=b    
        h=hashlib.md5(val.encode('utf-8')).hexdigest()
        if h[0:len(ans)]==ans:
            print(b+"\n"+h+"\n")
            f=open("a.md5",mode="a")
            f.write(b+"\n"+h+"\n")
            f.close()
            break
    print(" Process %d Finished." % n)
```

```python
def ser():
    for i in range(len(aaa)):
        for j in range(len(aaa)):
            for k in range(len(aaa)):
                for l in range(len(aaa)):
                    for m in range(len(aaa)):
                        b=aaa[i]+aaa[j]+aaa[k]+aaa[l]+aaa[m]
                        val+=b
                        h=hashlib.md5(val.encode('utf-8')).hexdigest()
                        if h[0:len(ans)]==ans:
                            print(b)
                            print(h)
```

```python
from multiprocessing import Pool
def pool_handler():
    with Pool(20) as p:
      p.map(ran, range(20))


if __name__ == '__main__':
    pool_handler()
```
