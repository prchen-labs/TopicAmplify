```python
from os import listdir
paths=["web crawler/files/{}/nchu-topic/",
       "web crawler/files/{}/nkust-topic/",
       "web crawler/files/{}/ntnu-topic/",
       "web crawler/files/{}/nuk-topic/",
      ]
topics = []
for opath in paths:
    path = opath.format("topics")
    for file in listdir(path):
        topics.append(opath + file) 
#topics=topics[:100]
```

```python
import pymongo

myclient = pymongo.MongoClient("mongodb://localhost:27017/")
mydb = myclient["topic_collection"]
myindex = mydb["indexs"]
myorig = mydb["original"]
myckip = mydb["ckip"]
indexs=[{"original_id":"","ckip_id":""} for _ in topics]
```

```python
x = myindex.delete_many({})
print(x.deleted_count, " documents deleted.") 
x = myorig.delete_many({})
print(x.deleted_count, " documents deleted.") 
x = myckip.delete_many({})
print(x.deleted_count, " documents deleted.") 
```

```python
import json
for i, opath in enumerate(topics):
    path = opath.format("topics")
    with open(path, mode="r") as f: 
        jsonn = json.loads(f.read())
        x = myorig.insert_one(jsonn)
        indexs[i]["original_id"]= x.inserted_id
    path = opath.format("ckip-topics")
    with open(path, mode="r") as f: 
        jsonn = json.loads(f.read())
        x = myckip.insert_one(jsonn)
        indexs[i]["ckip_id"] = x.inserted_id
```

```python
for index in indexs:
    x = myindex.insert_one(index)
print(len(indexs))
```
