```python
from ckipnlp.pipeline import CkipPipeline, CkipDocument
```

```python
import os
pipeline = CkipPipeline(opts={'con_parser': {'username': os.environ["CKIP_USERNAME"], 'password': os.environ["CKIP_PASSWORD"]}})
```

```python
info="""
【轉知】中華民國農科園區產學協會112年度企業人才培訓計畫課程規劃表
一、中華民國農科園區產學協會承攬行政院農業委員會屏東農業生物技術園區籌備處「農業生物科技園區112年度企業人才培訓計畫」（契約案號：111B24），依據契約書第二條辦理上揭活動。

 

二、年度課程規劃表、課程內容及授課地點詳如附件。

 

三、課程採線上報名，「新南向市場開發系列課程」免費，其餘課程收費請詳簡章，園區廠商及協會會員單位享優惠。

報名網址：http://www.rocaic.org/apply.php

聯絡電話：08-7629500

聯絡窗口：蘇育淋/傅榆婷

Email：redacted-user@example.com
"""
```

```python
doc = CkipDocument(raw=info)

# Word Segmentation
pipeline.get_ws(doc)
print(doc.ws)
for line in doc.ws:
    print(line.to_text())
```

```python
# Part-of-Speech Tagging
pipeline.get_pos(doc)
print(doc.pos)
for line in doc.pos:
    print(line.to_text())
```

```python
# Named-Entity Recognition
pipeline.get_ner(doc)
print(doc.ner)
```

```python
# Constituency Parsing
pipeline.get_conparse(doc)
print(doc.conparse)
```

```python
from ckipnlp.container.util.wspos import WsPosParagraph

# Word Segmentation & Part-of-Speech Tagging
for line in WsPosParagraph.to_text(doc.ws, doc.pos):
    print(line)
```
