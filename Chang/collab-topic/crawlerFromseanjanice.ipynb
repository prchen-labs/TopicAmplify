```python
import requests
import time

def make_request(url):
    try:
        response = requests.get(url)
        return response
    except:
        return None

# 設置重試次數
retry_count = 5

# 嘗試連接
while retry_count > 0:
    response = make_request("https://my.ntu.edu.tw/actregister/expiredActionList.aspx?year=2021")
    if response:
        # 連接成功，退出循環
        break
    retry_count -= 1
    time.sleep(5) # 設置等待時間

# 如果重試次數為 0，則表示連接失敗
if retry_count == 0:
    print("Error: Could not connect to the website.")
else:
    # 可以在這裡進一步處理 response 物件
    print(response.text)
```
