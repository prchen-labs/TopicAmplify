```python
from sklearn.feature_extraction.text import CountVectorizer

# 定義文本資料和標籤
corpus = ["This is a text about technology", "This is a text about politics", "This is a text about sports"]
labels = ["technology", "politics", "sports"]

# 建立CountVectorizer物件
vectorizer = CountVectorizer()

# 使用fit()方法對文本資料和標籤進行學習
vectorizer.fit(corpus, labels)

# 透過學習後可以獲得詞彙字典vocab
vocab = vectorizer.vocabulary_
print(vocab)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

# 定義文本資料和標籤
corpus = ["This is a text about technology", "This is a text about politics", "This is a text about sports"]
labels = ["technology", "politics", "sports"]

# 建立CountVectorizer物件
vectorizer = CountVectorizer()

# 使用fit()方法學習文本資料和標籤之間的關係
vectorizer.fit(corpus, labels)

# 建立分類器
classifier = MultinomialNB()

# 使用fit()方法訓練模型
classifier.fit(vectorizer.transform(corpus), labels)

# 預測新文本的類別
new_text = "This is a text about economy"
print("Predicted class:", classifier.predict(vectorizer.transform([new_text]))[0])
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# 定义文档数据
documents = [
    "This is the first document.",
    "This is the second document.",
    "And the third one.",
    "Is this the first document?"
]

# 创建 CountVectorizer 对象
vectorizer = CountVectorizer()
print("vectorizer: \n",vectorizer)

# 通过 fit_transform 方法构建词袋
bag_of_words = vectorizer.fit_transform(documents)

# 打印词袋矩阵
print("Bag of words matrix:\n", bag_of_words.toarray())

# 打印词汇表
print("Vocabulary:\n", vectorizer.vocabulary_)
```

詞袋模型（Bag of Words Model）是一种文本表示方法，它忽略了文本中单词的顺序和语法结构，**只考虑每个单词在文本中出现的频**率。它是基于词频的文本表示方法。

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer, CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation

# 加载数据
data = pd.read_csv("art_events.csv")
documents = data["Description"].values

# 建立 CountVectorizer 对象
vectorizer = CountVectorizer(stop_words='english')

# 建立词频矩阵
tf = vectorizer.fit_transform(documents)

# 建立 LDA 模型
lda_model = LatentDirichletAllocation(n_components=10, learning_method='online', random_state=0)

# 训练 LDA 模型
lda_model.fit(tf)

# 提取主题词语
feature_names = vectorizer.get_feature_names()
for topic_idx, topic in enumerate(lda_model.components_):
    print("Topic %d:" % (topic_idx + 1))
    print(" ".join([feature_names[i] for i in topic.argsort()[:-20 - 1:-1]]))
```

```python
from gensim import corpora, models

# 定义文档
documents = [["Human", "machine", "interface", "for", "lab", "abc", "computer", "applications"],
             ["A", "survey", "of", "user", "opinion", "of", "computer", "system", "response", "time"],
             ["The", "EPS", "user", "interface", "management", "system"],
             ["System", "and", "human", "system", "engineering", "testing", "of", "EPS"],              
             ["Relation", "of", "user", "perceived", "response", "time", "to", "error", "measurement"],
             ["The", "generation", "of", "random", "binary", "unordered", "trees"],
             ["The", "intersection", "graph", "of", "paths", "in", "trees"],
             ["Graph", "minors", "IV", "Widths", "of", "trees", "and", "well", "quasi", "ordering"],
             ["Graph", "minors", "A", "survey"]]
# 将文档转换为词袋
dictionary = corpora.Dictionary(documents)

# 将文档的词袋表示转换为词频矩阵
corpus = [dictionary.doc2bow(doc) for doc in documents]

# 训练 LDA 模型
ldamodel = models.LdaModel(corpus, num_topics=1, id2word=dictionary, passes=50)

# 输出主题
print("ldamodel.print_topics()[0]: ",ldamodel.print_topics()[0])
# print("ldamodel.print_topics()[1]: ",ldamodel.print_topics()[1])
# print("ldamodel.print_topics()[2]: ",ldamodel.print_topics()[2])

print(ldamodel.show_topics())
```
