```python
import requests
import json

url = 'https://cloud.culture.tw/frontsite/trans/SearchShowAction.do?method=doFindTypeJ&category=3'
response = requests.get(url)
data = json.loads(response.text)
# print(data)
```

```python
result = []
for item in data:
  title = item.get("title", None)
  show_info = item.get("showInfo", None)
  description_html = item.get("descriptionFilterHtml", None)
  image_url = item.get("imageUrl", None)
  master_unit = item.get("masterUnit", None)
  if master_unit == []:
    master_unit = None

  web_sales = item.get("webSales", None)
  source_web_promote = item.get("sourceWebPromote", None)

  for info in show_info:
      star_time = info.get("time", None)
      end_time = info.get("endTime", None)
      location = info.get("location", None)
      location_name = info.get("locationName", None)
      
      if web_sales == "":
          pay = "N"
      else:
          pay = "Y"
      
      result.append({
          "StartTime": star_time,
          "EndTime": end_time,
          "Place": location_name,
          "Address": location,
          "Title": title,
          "Pay": pay,
          "Info": source_web_promote,
          "Img": image_url,
          "Organizer": master_unit,
          "Speaker": None,
          "URL": web_sales,
          "Description": description_html,
          "Topic": "藝文活動",
          "Subcategory": "戲劇表演",
          "Source": "https://data.gov.tw/dataset/6015"
      })
# print(json.dumps(result, indent=4, ensure_ascii=False))
with open("/content/drive/MyDrive/CollectEventData/First舞蹈表演資訊.json", "w", encoding="utf-8") as f:
    json.dump(result, f, indent=4, ensure_ascii=False)
```

使用了 Python 中的字典(Dictionary)方法 get()。

字典的 get() 方法會在字典中查找給定的鍵 (key)，如果存在則返回對應的值，否則返回給定的預設值（在這裡是 None）。

例如：title = item.get("title", None)，在字典 item 中查找 key 為 "title" 的項，如果存在則返回對應的值，否則返回 None。

```python
import pandas as pd

# 讀取 csv 檔案並存入 DataFrame
df = pd.read_csv('/content/drive/MyDrive/CollectEventData/111年1月至8月文化部全國藝文活動填報資料_new (1).csv')

# 定義結果的字典
results = []
for index, row in df.iterrows():
    result = {}
    result['Title'] = row['活動名稱']
    date = row['日期'].split('～')
    result['StartTime'] = date[0]
    result['EndTime'] = date[1]
    result['Place'] = row['地點']
    
    if '演講／講座／研討會' in row['活動型態']:
        result['Topic'] = ['教育活動']
    elif '展覽' in row['活動型態']:
        result['Topic'] = ['藝文活動']
        result['Topic'] += ['商業活動']
    elif '表演與節慶' in row['活動型態']:
        result['Topic'] = ['藝文活動']
    elif '專題導覽' in row['活動型態']:
        result['Topic'] = ['藝文活動']
    elif '競賽／徵選與頒獎' in row['活動型態']:
        result['Topic'] = ['其他']
    elif '其他' in row['活動型態']:
        result['Topic'] = ['教育活動']
    
    result['Subcategory'] = row['活動型態']

    #未出現的數據，但我們需要
    result['Source'] = "https://data.gov.tw/dataset/157687"
    result['Address'] = ""
    result['Pay'] = ""
    result['Info'] = ""
    result['Img'] = ""
    result['Organizer'] = ""
    result['URL'] = ""
    result['Description'] = ""

    results.append(result)

# 將結果轉換為 json 格式並輸出
import json
print(json.dumps(results, ensure_ascii=False))
with open("/content/drive/MyDrive/CollectEventData/thirdF文化部全國藝文活動填報資料.json", "w", encoding="utf-8") as f:
    json.dump(results, f, indent=4, ensure_ascii=False)
```
